```python
import math
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

torch.manual_seed(42)


class LoRALinear(nn.Module):
    def __init__(self, base: nn.Linear, r: int = 4, alpha: int = 8):
        super().__init__()
        # Keep the original (pretrained) layer and freeze it
        self.base = base
        for p in self.base.parameters():
            p.requires_grad = False

        self.r = r
        self.alpha = alpha
        in_features = base.in_features
        out_features = base.out_features

        # A: shape (r, in_features), small random values
        self.A = nn.Parameter(torch.empty(r, in_features))
        nn.init.kaiming_uniform_(self.A, a=math.sqrt(5))

        # B: shape (out_features, r), zeros, so LoRA adds nothing at the start
        self.B = nn.Parameter(torch.zeros(out_features, r))

        self.scaling = alpha / r

    def forward(self, x):
        base_out = self.base(x)                    # (batch, out)
        lora_out = (x @ self.A.T) @ self.B.T       # (batch, r) -> (batch, out)
        return base_out + self.scaling * lora_out

    def merge(self):
        has_bias = self.base.bias is not None
        merged = nn.Linear(self.base.in_features, self.base.out_features, bias=has_bias,
                           device=self.base.weight.device, dtype=self.base.weight.dtype)
        with torch.no_grad():
            merged.weight.copy_(self.base.weight + self.scaling * (self.B @ self.A))
            if has_bias:
                merged.bias.copy_(self.base.bias)
        return merged
```

```python
layer = LoRALinear(nn.Linear(16, 16), r=4, alpha=8)
for name, p in layer.named_parameters():
    print(name, p.shape, p.requires_grad)
```

```python
layer = LoRALinear(nn.Linear(16, 16), r=4, alpha=8)
x = torch.randn(5, 16)
print(torch.allclose(layer(x), layer.base(x)))   # True
```

```python
torch.manual_seed(0)
# The "pretrained" layer (frozen inside LoRA) and a new "target" behaviour to learn
layer = LoRALinear(nn.Linear(16, 16), r=4, alpha=8)
target = nn.Linear(16, 16)

# Training data: the target must be a SMALL change of the base layer
# (that's the real fine-tuning situation: we adjust, we don't relearn everything)
with torch.no_grad():
    target.weight.copy_(layer.base.weight + 0.1 * torch.randn(16, 4) @ torch.randn(4, 16))
    target.bias.copy_(layer.base.bias)

X = torch.randn(1000, 16)
y = target(X).detach()

optimizer = torch.optim.Adam([
    p for p in layer.parameters() if p.requires_grad
], lr = 1e-2)

loss_fn = nn.MSELoss()

losses = []
for step in range(300):
  pred = layer(X)
  loss = loss_fn(pred , y)
  optimizer.zero_grad()
  loss.backward()
  optimizer.step()
  losses.append(loss.item())
  if step % 50 == 0 :
    print(f"step {step:3d}  loss {loss.item():.6f}")

```

```python
import matplotlib.pyplot as plt
plt.plot(losses)
plt.xlabel("step"); plt.ylabel("MSE loss"); plt.yscale("log")
plt.title("LoRA learns the difference between base and target")
plt.show()
```

```python
import json
from huggingface_hub import hf_hub_download

def load_json(filename):
    path = hf_hub_download("cat-overflow/FormulaReasoning", filename, repo_type="dataset")
    with open(path, encoding="utf-8") as f:
        try:
            return json.load(f)                         # a regular .json file
        except json.JSONDecodeError:
            f.seek(0)
            return [json.loads(line) for line in f]     # a .jsonl file (one line = one example)

hef_test = load_json("formulareasoning/HeF_test.json")

print(type(hef_test), len(hef_test))
example = hef_test[0] if isinstance(hef_test, list) else hef_test
print(json.dumps(example, ensure_ascii=False, indent=2)[:2000])
```

```python
import re, random
train_raw = load_json("formulareasoning/train.json")
hof_test  = load_json("formulareasoning/HoF_test.json")
hef_test  = load_json("formulareasoning/HeF_test.json")
print(f"train: {len(train_raw)} | HoF test: {len(hof_test)} | HeF test: {len(hef_test)}")
```

```python
def clean(name):
    # "[ heat absorbed by water] " -> "heat absorbed by water"
    return " ".join(name.strip().strip("[]").split()).lower()

def build_solution(ex):
    # dictionary: quantity name -> "value unit"
    values = {clean(a["en_name"]): f'{a["num"]} {a["unit"]}'.strip() for a in ex["arguments"]}
    lines = []
    for i, step in enumerate(ex["solution"]["en"], 1):
        lhs = step.split("=")[0]                  # left side of the formula
        line = f"Step {i}: {step.strip()}"
        val = values.get(clean(lhs))
        if val:                                    # add the computed result if we have it
            line += f"\n  => {lhs.strip()} = {val}"
        lines.append(line)
    lines.append(f"Answer: {ex['answer']}")
    return "\n".join(lines)

print(train_raw[0]["question"]["en"])
print("-----")
print(build_solution(train_raw[0]))
```

```python
def parse_number(s):
    if s is None:
        return None
    s = s.replace(",", "").replace("×", "*").replace(" ", "")
    m = re.search(r"(-?\d+(?:\.\d+)?)(?:\*10\^\{?\(?(-?\d+)\)?\}?|[eE](-?\d+))?", s)
    if not m:
        return None
    value = float(m.group(1))
    exp = m.group(2) or m.group(3)
    if exp:
        value *= 10 ** int(exp)
    return value

# Test: how many answers in train can we read?
failed = [ex["answer"] for ex in train_raw if parse_number(ex["answer"]) is None]
print(f"unreadable: {len(failed)} / {len(train_raw)}")
print(failed[:10])
print(parse_number("1.4×10^8 J/kg"), parse_number("0.3 kg"), parse_number("4.2*10^3"))
```

```python
SYSTEM = ("You are a physics expert. Solve the problem step by step using physics formulas. "
          "End with: Answer: <number> <unit>")

def make_messages(ex, with_solution):
    msgs = [{"role": "system", "content": SYSTEM},
            {"role": "user", "content": ex["question"]["en"]}]
    if with_solution:
        msgs.append({"role": "assistant", "content": build_solution(ex)})
    return msgs

def tokenize_example(ex):
    # prompt = system + question (what the model receives)
    prompt = tokenizer.apply_chat_template(make_messages(ex, False), tokenize=False, add_generation_prompt=True)
    # full = prompt + solution (what the model must learn to write)
    full = tokenizer.apply_chat_template(make_messages(ex, True), tokenize=False)

    prompt_ids = tokenizer(prompt, add_special_tokens=False)["input_ids"]
    full_ids   = tokenizer(full,   add_special_tokens=False)["input_ids"]
    assert full_ids[:len(prompt_ids)] == prompt_ids   # the prompt must be the start of the full text

    # -100 = "don't learn this": the model is only graded on the solution
    labels = [-100] * len(prompt_ids) + full_ids[len(prompt_ids):]
    return {"input_ids": full_ids, "labels": labels}

MAX_LEN = 1024
train_data = [tokenize_example(ex) for ex in train_raw]
train_data = [d for d in train_data if len(d["input_ids"]) <= MAX_LEN]
lengths = [len(d["input_ids"]) for d in train_data]
print(f"examples kept: {len(train_data)} | average length: {sum(lengths)/len(lengths):.0f} tokens | max: {max(lengths)}")
```

```python
random.seed(42)
train_small = random.sample(train_data, 200)
hof_small = hof_test[:50]
hef_small = hef_test[:50]
```

```python
def extract_answer(text):
    # 1. Look for the LAST "Answer:" in the text
    matches = re.findall(r"Answer:\s*(.+)", text)
    if matches:
        return parse_number(matches[-1])
    # 2. Otherwise, take the last number in the text (fallback)
    numbers = re.findall(r"-?\d+(?:\.\d+)?(?:\s*[×*]\s*10\^\{?\(?-?\d+\)?\}?)?", text)
    return parse_number(numbers[-1]) if numbers else None

def is_correct(pred, gold, tol=0.01):
    if pred is None or gold is None:
        return False
    if gold == 0:
        return abs(pred) < 1e-6
    return abs(pred - gold) / abs(gold) < tol      # relative error < 1%

# Quick tests
print(extract_answer("Step 1: ...\nAnswer: 0.3 kg"))          # 0.3
print(extract_answer("so Q = 4.2×10^7 J"))                   # 42000000.0
print(is_correct(0.301, 0.3), is_correct(0.35, 0.3))        # True False
```

```python
tokenizer.padding_side = "left"   # for generation, padding goes on the LEFT

@torch.no_grad()
def generate_answers(model, examples, batch_size=8, max_new_tokens=512):
    model.eval()
    outputs = []
    for i in range(0, len(examples), batch_size):
        batch = examples[i:i + batch_size]
        prompts = [tokenizer.apply_chat_template(make_messages(ex, False), tokenize=False,
                                                 add_generation_prompt=True) for ex in batch]
        inputs = tokenizer(prompts, return_tensors="pt", padding=True, add_special_tokens=False).to(device)
        out = model.generate(**inputs, max_new_tokens=max_new_tokens, do_sample=False,
                             pad_token_id=tokenizer.pad_token_id)
        for seq in out:
            outputs.append(tokenizer.decode(seq[inputs["input_ids"].shape[1]:], skip_special_tokens=True))
        print(f"  {min(i + batch_size, len(examples))}/{len(examples)}")
    return outputs
```

```python
def evaluate(model, examples, name):
    texts = generate_answers(model, examples)
    records, correct, formatted = [], 0, 0
    for ex, text in zip(examples, texts):
        gold = parse_number(ex["answer"])
        pred = extract_answer(text)
        ok = is_correct(pred, gold)
        correct += ok
        formatted += "Answer:" in text
        records.append({"id": ex["id"], "gold": ex["answer"], "pred": pred, "correct": ok, "output": text})
    acc = correct / len(examples)
    print(f"{name}: accuracy = {acc:.1%} | 'Answer:' format respected = {formatted/len(examples):.0%}")
    return acc, records
```

```python
base_hof_acc, base_hof_records = evaluate(model, hof_small, "BASE - HoF")
base_hef_acc, base_hef_records = evaluate(model, hef_small, "BASE - HeF")

# Look at one answer, to understand the mistakes
r = base_hof_records[0]
print("\nGOLD:", r["gold"], "| PRED:", r["pred"], "| correct:", r["correct"])
print(r["output"][:1500])
```

```python
test_ids = torch.tensor([train_small[0]["input_ids"]]).to(device)
model.eval()
with torch.no_grad():
    logits_before = model(test_ids).logits
```

```python
TARGETS = ["q_proj", "k_proj", "v_proj", "o_proj"]
R, ALPHA = 8, 16

# 1. Geler TOUS les poids de Qwen
for p in model.parameters():
    p.requires_grad = False

# 2. Trouver les couches à remplacer (on fait la liste AVANT de modifier le modèle)
to_replace = [(name, module) for name, module in model.named_modules()
              if name.split(".")[-1] in TARGETS and isinstance(module, nn.Linear)]

# 3. Remplacer chaque couche par ton LoRALinear
for name, module in to_replace:
    parent_name, child_name = name.rsplit(".", 1)      # "model.layers.0.self_attn", "q_proj"
    parent = model.get_submodule(parent_name)
    setattr(parent, child_name, LoRALinear(module, r=R, alpha=ALPHA).to(device))

print(f"couches remplacées : {len(to_replace)}")       # 24 couches × 4 = 96
print(model.model.layers[0].self_attn)                 # tu dois voir LoRALinear
```

```python
# Vérification 1 : la sortie n'a pas changé (B = 0)
with torch.no_grad():
    logits_after = model(test_ids).logits
print("Sortie identique :", torch.allclose(logits_before, logits_after, atol=1e-4))

# Vérification 2 : combien de paramètres on entraîne
trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
total = sum(p.numel() for p in model.parameters())
print(f"Entraînables : {trainable:,} / {total:,} = {100*trainable/total:.2f}%")
```

```python
from torch.utils.data import DataLoader

def collate(batch):
    # Les exemples n'ont pas la même longueur : on complète (padding) jusqu'au plus long
    max_len = max(len(d["input_ids"]) for d in batch)
    pad = tokenizer.pad_token_id
    input_ids, labels, attention = [], [], []
    for d in batch:
        n = max_len - len(d["input_ids"])
        input_ids.append(d["input_ids"] + [pad] * n)
        labels.append(d["labels"] + [-100] * n)            # le padding n'est pas appris
        attention.append([1] * len(d["input_ids"]) + [0] * n)  # 0 = "ignore ce token"
    return {"input_ids": torch.tensor(input_ids),
            "labels": torch.tensor(labels),
            "attention_mask": torch.tensor(attention)}

BATCH_SIZE = 2
loader = DataLoader(train_small, batch_size=BATCH_SIZE, shuffle=True, collate_fn=collate)
```

```python
from transformers import get_linear_schedule_with_warmup

EPOCHS = 3
ACCUM = 4          # on additionne 4 batches avant chaque mise à jour → batch effectif = 8
LR = 2e-4

trainable_params = [p for p in model.parameters() if p.requires_grad]   # seulement les A et B
optimizer = torch.optim.AdamW(trainable_params, lr=LR)

total_updates = EPOCHS * len(loader) // ACCUM
scheduler = get_linear_schedule_with_warmup(optimizer,
                                            num_warmup_steps=int(0.1 * total_updates),
                                            num_training_steps=total_updates)
print(f"Nombre de mises à jour : {total_updates}")
```

```python
import time
model.train()
losses = []
step = 0
start = time.time()

for epoch in range(EPOCHS):
    for i, batch in enumerate(loader):
        batch = {k: v.to(device) for k, v in batch.items()}

        outputs = model(**batch)                  # 1. forward (la loss est calculée par Qwen)
        loss = outputs.loss / ACCUM               # 2. erreur (divisée car on accumule)
        loss.backward()                           # 3. gradients de A et B

        if (i + 1) % ACCUM == 0:
            torch.nn.utils.clip_grad_norm_(trainable_params, 1.0)   # évite les gradients explosifs
            optimizer.step()                      # 4. met à jour A et B
            scheduler.step()
            optimizer.zero_grad()                 # 5. remet les gradients à zéro
            step += 1
            losses.append(loss.item() * ACCUM)
            if step % 5 == 0:
                print(f"epoch {epoch} | update {step}/{total_updates} | loss {losses[-1]:.4f}")

print(f"Terminé en {(time.time() - start)/60:.1f} min")
print(f"Mémoire GPU max : {torch.cuda.max_memory_allocated()/1e9:.1f} GB")
```

```python
plt.plot(losses)
plt.xlabel("mise à jour"); plt.ylabel("loss")
plt.title("Fine-tuning LoRA sur FormulaReasoning")
plt.show()
```

## Why LoRA is still the better choice, even with similar accuracy

Even when LoRA and full fine-tuning reach similar accuracy, LoRA wins on cost. It trains only
~1.08M parameters (≈0.22% of Qwen2.5-0.5B) while the base weights W₀ stay frozen. That cuts
GPU memory enough to fine-tune on a free Colab T4, and it produces a ~4 MB adapter instead of
a ~2 GB model copy.
After `merge()` (W = W₀ + (α/r)·B·A), inference is exactly as fast as the
original model. And because the low-rank update can only change the model in a few directions,
LoRA keeps the model's general knowledge and limits overfitting, which is what the HeF test
(unseen formulas) lets us check.
Same result, for a fraction of the cost.
